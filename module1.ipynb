```python
# ==========================================
# OBJECTIVE 1 & 2: Environment Verification & Python Basics for Data Science
# ==========================================
import sys
import json
import csv

print(f"Python Version: {sys.version.split()[0]} (Google Colab Environment Ready)\n")

# Variables & Basic Data Types (Data Science Metrics)
project_name = "Customer Churn Risk Analysis"
total_records = 5
churn_threshold = 0.65
is_production_ready = True

print(f"Project: {project_name} | Threshold: {churn_threshold} | Active: {is_production_ready}\n")


# ==========================================
# OBJECTIVE 3: Lists, Dictionaries, Loops, and Functions
# ==========================================
# List of dictionaries representing customer usage data
customers = [
    {"id": "C101", "name": "Aarav", "monthly_spend": 120.50, "support_tickets": 1, "tenure_months": 24},
    {"id": "C102", "name": "Diya", "monthly_spend": 45.00, "support_tickets": 5, "tenure_months": 3},
    {"id": "C103", "name": "Rohan", "monthly_spend": 89.99, "support_tickets": 2, "tenure_months": 14},
    {"id": "C104", "name": "Sneha", "monthly_spend": 30.00, "support_tickets": 6, "tenure_months": 2},
    {"id": "C105", "name": "Vikram", "monthly_spend": 150.00, "support_tickets": 0, "tenure_months": 36},
]

def calculate_churn_score(customer: dict) -> float:
    """Calculates a heuristic churn risk score between 0.0 and 1.0."""
    ticket_weight = min(customer["support_tickets"] * 0.12, 0.60)
    tenure_discount = min(customer["tenure_months"] * 0.015, 0.40)
    score = round(max(0.05, 0.45 + ticket_weight - tenure_discount), 2)
    return score

# Loop through records to compute scores and classify risk
high_risk_customers = []
for cust in customers:
    cust["churn_score"] = calculate_churn_score(cust)
    cust["risk_level"] = "HIGH" if cust["churn_score"] >= churn_threshold else "LOW"
    if cust["risk_level"] == "HIGH":
        high_risk_customers.append(cust["name"])

print("Processed Customer Records:")
for c in customers:
    print(f" - {c['id']} ({c['name']}): Spend=${c['monthly_spend']} | Score={c['churn_score']} [{c['risk_level']}]")
print(f"\nHigh-Risk Customers Flagged: {high_risk_customers}\n")


# ==========================================
# OBJECTIVE 3 (Cont.): File Handling (CSV & JSON Read/Write)
# ==========================================
csv_filename = "customer_churn_report.csv"
json_filename = "summary_metrics.json"

# 1. Write processed dataset to a CSV file
fieldnames = ["id", "name", "monthly_spend", "support_tickets", "tenure_months", "churn_score", "risk_level"]
with open(csv_filename, mode="w", newline="") as file:
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(customers)

# 2. Read back the CSV file to verify persistence
with open(csv_filename, mode="r") as file:
    reader = list(csv.DictReader(file))
    print(f"Successfully wrote and verified {len(reader)} rows in '{csv_filename}'.")

# 3. Save summary analytics to a JSON file
summary_data = {
    "project": project_name,
    "total_customers": len(customers),
    "avg_monthly_spend": round(sum(c["monthly_spend"] for c in customers) / len(customers), 2),
    "high_risk_count": len(high_risk_customers),
    "high_risk_names": high_risk_customers
}

with open(json_filename, mode="w") as jfile:
    json.dump(summary_data, jfile, indent=4)

# Read JSON back
with open(json_filename, mode="r") as jfile:
    loaded_summary = json.load(jfile)
    print("Loaded JSON Summary:", loaded_summary)
```
