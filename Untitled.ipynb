```python
# ==========================================
# OBJECTIVE 1: Setup Matplotlib & Seaborn
# ==========================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set clean visual theme
sns.set_theme(style="whitegrid", palette="viridis")
plt.rcParams["figure.dpi"] = 120

# Create a rich SaaS Customer Telemetry dataset for visualization
np.random.seed(42)
n_customers = 120

plans = np.random.choice(["Basic", "Pro", "Enterprise"], size=n_customers, p=[0.45, 0.35, 0.20])
tenure = np.random.randint(1, 37, size=n_customers)

# Generate realistic monthly spend & support tickets correlated with plan and tenure
base_spend = np.where(plans == "Basic", 45, np.where(plans == "Pro", 110, 260))
monthly_spend = np.round(base_spend + (tenure * 1.2) + np.random.normal(0, 12, size=n_customers), 2)
support_tickets = np.clip(np.round(6 - (tenure * 0.12) + np.random.normal(0, 1.5, size=n_customers)), 0, 10).astype(int)

df = pd.DataFrame({
    "Customer_ID": [f"C{1000+i}" for i in range(n_customers)],
    "Plan_Type": plans,
    "Tenure_Months": tenure,
    "Monthly_Spend": monthly_spend,
    "Support_Tickets": support_tickets
})

# Simulated 6-month revenue growth trend for Line Chart
months = ["Jun", "Jul", "Aug", "Sep", "Oct", "Nov"]
monthly_revenue = [11200, 12450, 13100, 14800, 16250, 18400]


# ==========================================
# OBJECTIVE 2: Build Bar, Line, Pie, Histogram & Scatter Plots
# ==========================================
fig = plt.figure(figsize=(16, 10))
fig.suptitle("Customer Telemetry & Revenue Analytics Dashboard", fontsize=16, fontweight="bold", y=0.98)

# 1. BAR CHART: Average Monthly Spend by Subscription Plan (Seaborn)
ax1 = plt.subplot(2, 3, 1)
sns.barplot(
    data=df, x="Plan_Type", y="Monthly_Spend",
    order=["Basic", "Pro", "Enterprise"], estimator=np.mean, errorbar=None, ax=ax1, hue="Plan_Type", legend=False
)
ax1.set_title("1. Avg Monthly Spend by Plan (Bar Chart)", fontweight="bold")
ax1.set_xlabel("Subscription Plan")
ax1.set_ylabel("Avg Monthly Spend ($)")
for container in ax1.containers:
    ax1.bar_label(container, fmt="$%.0f", padding=3)

# 2. LINE CHART: 6-Month Recurring Revenue Trend (Matplotlib)
ax2 = plt.subplot(2, 3, 2)
ax2.plot(months, monthly_revenue, marker="o", linewidth=2.5, markersize=8, color="#2b5c8f")
ax2.fill_between(months, monthly_revenue, color="#2b5c8f", alpha=0.12)
ax2.set_title("2. 6-Month Revenue Growth (Line Chart)", fontweight="bold")
ax2.set_xlabel("Month")
ax2.set_ylabel("Total Revenue ($)")
for i, val in enumerate(monthly_revenue):
    ax2.annotate(f"${val:,}", (months[i], val), textcoords="offset points", xytext=(0, 8), ha="center", fontsize=8.5)

# 3. PIE CHART: Customer Distribution by Plan Tier (Matplotlib)
ax3 = plt.subplot(2, 3, 3)
plan_counts = df["Plan_Type"].value_counts()
colors = sns.color_palette("viridis", len(plan_counts))
ax3.pie(
    plan_counts, labels=plan_counts.index, autopct="%1.1f%%",
    startangle=140, colors=colors, wedgeprops={"edgecolor": "white", "linewidth": 1.5}
)
ax3.set_title("3. Customer Share by Plan (Pie Chart)", fontweight="bold")

# 4. HISTOGRAM: Distribution of Customer Tenure (Seaborn)
ax4 = plt.subplot(2, 3, 4)
sns.histplot(data=df, x="Tenure_Months", bins=10, kde=True, color="#2a9d8f", ax=ax4)
ax4.axvline(df["Tenure_Months"].mean(), color="crimson", linestyle="--", linewidth=1.5, label=f"Mean: {df['Tenure_Months'].mean():.1f} mos")
ax4.set_title("4. Customer Tenure Distribution (Histogram)", fontweight="bold")
ax4.set_xlabel("Tenure (Months)")
ax4.set_ylabel("Customer Count")
ax4.legend()

# 5. SCATTER PLOT: Tenure vs. Monthly Spend by Plan (Seaborn)
ax5 = plt.subplot(2, 3, (5, 6))
sns.scatterplot(
    data=df, x="Tenure_Months", y="Monthly_Spend",
    hue="Plan_Type", style="Plan_Type", s=75, alpha=0.85, ax=ax5
)
ax5.set_title("5. Tenure vs. Monthly Spend Across Plans (Scatter Plot)", fontweight="bold")
ax5.set_xlabel("Tenure (Months)")
ax5.set_ylabel("Monthly Spend ($)")
ax5.legend(title="Plan Type", loc="upper left")

plt.tight_layout(rect=[0, 0, 1, 0.94])
plt.savefig("module3_visualizations.png", dpi=150)
plt.show()


# ==========================================
# OBJECTIVE 3: Identify Patterns, Trends & Key Insights
# ==========================================
print("\n=== KEY PATTERNS & DATA INSIGHTS ===")
rev_growth = ((monthly_revenue[-1] - monthly_revenue[0]) / monthly_revenue[0]) * 100
corr_tenure_spend = df["Tenure_Months"].corr(df["Monthly_Spend"])
corr_tenure_tickets = df["Tenure_Months"].corr(df["Support_Tickets"])

print(f"1. Revenue Trajectory (Line Chart): Monthly recurring revenue grew by {rev_growth:.1f}% from Jun (${monthly_revenue[0]:,}) to Nov (${monthly_revenue[-1]:,}).")
print(f"2. Tier Monetization (Bar & Pie Charts): While Basic accounts for the largest user share ({plan_counts['Basic']/len(df)*100:.1f}%), Enterprise users spend over 5x more per month on average.")
print(f"3. Retention & Expansion (Scatter Plot & Histogram): Longer customer tenure correlates positively with monthly spend (r = {corr_tenure_spend:.2f}) and negatively with support tickets (r = {corr_tenure_tickets:.2f}), showing that long-term customers require less support while generating higher revenue.")
```
